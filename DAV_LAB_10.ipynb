```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/content/drive/MyDrive/archive (10)/players_21.csv")

df['Primary_Position'] = df['player_positions'].apply(lambda x: x.split(',')[0].strip())

position_counts = df['Primary_Position'].value_counts()

plt.figure(figsize=(10,5))
plt.bar(position_counts.index, position_counts.values, color='skyblue')
plt.xticks(rotation=90)
plt.xlabel("Player Position")
plt.ylabel("Number of Players")
plt.title("Number of Players Across Primary Positions")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/content/drive/MyDrive/archive (10)/players_21.csv")

df['Primary_Position'] = df['player_positions'].apply(lambda x: x.split(',')[0].strip())

# Rating ranges
bins = [40, 60, 70, 80, 90, 100]
labels = ["40-60", "60-70", "70-80", "80-90", "90-100"]

df["RatingRange"] = pd.cut(df["overall"], bins=bins, labels=labels)

# Count by rating range
rating_counts = df["RatingRange"].value_counts().sort_index()

# Donut Plot
plt.figure(figsize=(8,8))
plt.pie(rating_counts, labels=labels, autopct="%1.1f%%", startangle=90)

# Donut effect
centre_circle = plt.Circle((0,0), 0.70, color="white")
plt.gca().add_artist(centre_circle)

plt.title("Distribution of Player Overall Ratings by Range")
plt.show()
```
